# Polymarket loss versus win follow-up

This notebook inspects the 159,666 wallets eligible for the first loss-chasing pass. It compares each wallet's average next-day stake after loss days with average next-day stake after win days. The 2026 trade holdout stays untouched.

Source documentation: https://www.vincentgregoire.com/polymarket-users-data/

```python
import json
from pathlib import Path
import polars as pl

OUTPUT = Path('analysis/output')
summary = json.loads((OUTPUT / 'loss-vs-win-summary.json').read_text())
comparison = pl.read_parquet(OUTPUT / 'loss-win-wallet-comparison.parquet')
chasing = pl.read_csv(OUTPUT / 'consistent-chasing-wallets.csv')
summary['population'], summary['all_wallet_average']
```

```python
# Category distribution across all 159,666 first-pass eligible wallets.
comparison.group_by('behavior_category').len().with_columns(
    (pl.col('len') / comparison.height).alias('share_all_eligible')
).sort('behavior_category')
```

```python
# Inspect the 1,050 repeated-chasing wallets.
chasing.select([
    'user_address', 'loss_episode_count', 'chasing_episode_count',
    'chasing_episode_share', 'median_post_loss_ratio',
    'baseline_median_stake_usd', 'win_mean_post_stake_usd',
    'win_mean_post_ratio', 'behavior_category'
]).head(20)
```

Interpretation: the median wallet shows a small post-loss reduction. Raw-dollar averages also show lower post-loss stake. The normalized mean is much larger because a small number of wallets have extreme ratios against small baselines, so median and raw-dollar results deserve more weight.
